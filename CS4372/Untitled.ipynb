```python
df = pd.read_csv("https://raw.githubusercontent.com/a-nagar/cs4372/refs/heads/main/data/roc_disease.csv")
df
```

```python
TP = df['actual'].value_counts()[1]
TN_FP = df['actual'].value_counts()[0]
((df['pred_prob'] > 0.5) & (df['actual'] == 1)).sum()/TP
((df['pred_prob'] > 0.5) & (df['actual'] == 0)).sum()/TN_FP
```

```python
lattice = np.linspace(0, 1, num = 50)
FPRs = []
TPRs = []
for i in lattice: 
    FPRs.append(((df['pred_prob'] > i) & (df['actual'] == 0)).sum()/TN_FP)
    TPRs.append(((df['pred_prob'] > i) & (df['actual'] == 1)).sum()/TP)
```

```python
FPR_tuples = zip(FPRs[1:],FPRs[0:-1])
```

```python
AUC = sum([tpr * (fpr[1] - fpr[0]) for tpr, fpr in zip(TPRs, FPR_tuples)])
```
